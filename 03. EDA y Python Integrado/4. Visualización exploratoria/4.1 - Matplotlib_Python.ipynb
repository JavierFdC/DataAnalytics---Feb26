# **Matplotlib**
---

Matplotlib es una biblioteca de visualización 2D en Python que permite crear gráficos y visualizaciones de datos de manera sencilla y flexible. Fue desarrollada originalmente por John D. Hunter en 2003 y desde entonces ha evolucionado para convertirse en una de las herramientas más confiables y versátiles para la creación de gráficos en Python.

<span style="color:orange"> **Importancia de Matplotlib:** </span>

1. **Facilidad de uso:** Matplotlib es fácil de aprender y usar, lo que la convierte en una elección popular tanto para principiantes como para usuarios avanzados en Python. La API de estilo Pyplot facilita la creación rápida de gráficos simples.

2. **Calidad de visualización:** Matplotlib produce gráficos de alta calidad con una amplia gama de opciones de personalización. Puedes controlar detalles como colores, estilos de línea, etiquetas, títulos y más, lo que te permite crear visualizaciones que se adapten a tus necesidades específicas.

3. **Soporte de múltiples formatos de salida:** Puedes guardar tus gráficos en diversos formatos, como PNG, JPEG, PDF, SVG y otros. Esto es útil para la inclusión de gráficos en informes, presentaciones y publicaciones.

4. **Amplia gama de tipos de gráficos:** Matplotlib admite una variedad de tipos de gráficos, incluyendo gráficos de líneas, de barras, de dispersión, de pastel, de histograma, de cajas, de superficie, y más. Puedes elegir el tipo de gráfico que mejor se adapte a tus datos y a tus objetivos de visualización.

5. **Personalización avanzada:** Además de las opciones básicas de personalización, Matplotlib permite un alto grado de control y ajuste de tus gráficos. Puedes personalizar la disposición de ejes, añadir anotaciones, barras de error y más.

6. **Soporte de gráficos 3D:** Matplotlib también es capaz de crear visualizaciones 3D, lo que es útil en campos como la visualización científica y la representación de datos tridimensionales.

7. **Integración con otros paquetes:** Matplotlib se integra bien con otras bibliotecas y herramientas de Python utilizadas en análisis de datos, como NumPy, Pandas y SciPy.

8. **Documentación y comunidad activa:** Matplotlib cuenta con una amplia documentación oficial y una comunidad activa de usuarios y desarrolladores que proporcionan soporte, ejemplos y soluciones a preguntas comunes.

# **Introducción**

* <span style="color:orange">**Figuras y ejes:** </span>

Matplotlib organiza visualizaciones en dos componentes principales: figuras (figures) y ejes (axes). Una figura es la ventana gráfica en la que se muestra el gráfico, y los ejes son los ejes X e Y donde se dibuja el gráfico.

* <span style="color:orange">**Crear gráficos básicos:** </span>

Puedes crear gráficos de líneas, de dispersión, de barras, de pastel y muchos otros tipos de gráficos utilizando Matplotlib. Aquí tienes un ejemplo simple de un gráfico de líneas:

```python
x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 18, 20]

plt.plot(x, y)
plt.title('Gráfico de Líneas')
plt.xlabel('Eje X')
plt.ylabel('Eje Y')
plt.show()
```

* <span style="color:orange">**Personalización de gráficos:** </span>

Matplotlib te permite personalizar tus gráficos de muchas maneras, como cambiar colores, estilos de línea, agregar etiquetas, títulos, leyendas, etc. Puedes ajustar la apariencia de tus gráficos según tus necesidades.

* <span style="color:orange"> **Guardar gráficos:** </span>

Puedes guardar tus gráficos en diferentes formatos, como PNG, JPEG, PDF, SVG, etc., utilizando **plt.savefig()**.

```python
plt.savefig('mi_grafico.pdf')
```

* <span style="color:orange"> **Subplots:** </span>

Puedes crear múltiples gráficos en una sola figura utilizando subplots. Esto es útil para comparar diferentes conjuntos de datos o mostrar múltiples gráficos en una sola imagen.

```python
x1 = [1, 2, 3, 4, 5]
y1= [10, 15, 13, 18, 20]

x2 = [2, 4, 6, 8, 10]
y2 = [20, 25, 34, 12, 11]

plt.subplot(2, 2, 1)  # Crear un subplot en la posición 1
plt.plot(x1, y1)

plt.subplot(2, 2, 2)  # Crear un subplot en la posición 2
plt.plot(x2, y2)

# Puedes crear más subplots según sea necesario
```

###  <span style="color:lightgreen"> **Gráfico de lineas**  </span>

Un gráfico de líneas, es un tipo de representación gráfica que muestra la relación entre dos variables a lo largo de un sistema de coordenadas cartesianas. En un gráfico de líneas, los puntos de datos se conectan mediante líneas rectas, lo que permite visualizar cómo cambian los valores de una variable (generalmente en el eje Y) en función de los valores de otra variable (generalmente en el eje X).

**La importancia de los gráficos de líneas radica en varias razones:**

1. **Visualización de tendencias:** Los gráficos de líneas son ideales para visualizar tendencias y patrones en datos. Pueden ayudar a identificar cambios en los valores a lo largo del tiempo o en función de otra variable, lo que es esencial en análisis de datos y toma de decisiones.

2. **Comunicación efectiva:** Son una forma efectiva de comunicar información a través de una representación visual clara y fácil de entender. Esto es útil en informes, presentaciones y publicaciones para ilustrar hallazgos y resultados.

3. **Comparación de series de datos:** Los gráficos de líneas permiten comparar múltiples series de datos en un solo gráfico, lo que facilita la comparación de tendencias y relaciones entre diferentes conjuntos de datos.

4. **Análisis de datos de series temporales:** Son fundamentales en el análisis de datos de series temporales, como seguimiento de valores a lo largo del tiempo en campos como las finanzas, la meteorología, la epidemiología y otros.

5. **Identificación de valores atípicos:** Los valores atípicos o anómalos a menudo son visibles en un gráfico de líneas y pueden ser detectados de manera más efectiva que en otros tipos de gráficos.

6. **Facilitan la extrapolación e interpolación:** Los gráficos de líneas permiten estimar valores entre puntos de datos conocidos, lo que es útil en predicciones y análisis.

La ubicación (**loc**) es un argumento que acepta varios valores para controlar dónde aparece la leyenda en el gráfico. Algunos ejemplos de ubicaciones comunes que puedes especificar incluyen:

* **'upper right':** Coloca la leyenda en la esquina superior derecha del gráfico.
* **'upper left':** Coloca la leyenda en la esquina superior izquierda del gráfico.
* **'lower right':** Coloca la leyenda en la esquina inferior derecha del gráfico (como en tu código).
* **'lower left':** Coloca la leyenda en la esquina inferior izquierda del gráfico.


También puedes utilizar otros valores como **'upper center', 'center', 'center left', 'center right'**, entre otros, para controlar la posición exacta de la leyenda en el gráfico.

 <span style="color:orange"> **Intervalos** </span>

Los intervalos o límites en un gráfico son útiles por varias razones:

1. **Enfocar en datos específicos:** Al establecer límites en los ejes, puedes enfocarte en un rango específico de datos de interés. Esto te permite examinar con mayor detalle un período de tiempo o un conjunto de valores particulares en tus datos.

2. **Resaltar tendencias o patrones:** Los límites pueden ayudarte a resaltar tendencias o patrones que pueden no ser tan evidentes si se muestran todos los datos. Al acercar la vista a un intervalo específico, puedes identificar tendencias, fluctuaciones o eventos destacados en ese rango.

3. **Reducir el ruido visual:** En conjuntos de datos grandes, mostrar todos los datos en un solo gráfico puede generar un exceso de información visual. Al limitar los ejes a un intervalo relevante, puedes reducir el ruido visual y centrarte en lo que es más importante.

4. **Comparación de datos:** Puedes utilizar límites en ejes para comparar diferentes segmentos de tus datos en el mismo gráfico. Esto es útil cuando deseas mostrar cómo dos o más series de datos se comportan en un intervalo de tiempo específico.

5. **Mejora la legibilidad:** En gráficos con escalas extensas, como escalas de tiempo, los límites pueden ayudar a mantener el gráfico legible al mostrar solo la porción de datos que es relevante en un momento dado.

6. **Enfatizar eventos importantes:** Al ajustar los límites para resaltar un evento o un punto específico en tus datos, puedes llamar la atención sobre ese evento y destacarlo en tu gráfico.

```python
# Importar la librería datetime
import datetime

# Crear una figura con un tamaño específico
fig = plt.figure(figsize=(15, 8))

# Agregar un conjunto de ejes que ocupe toda la figura
eje = fig.add_axes([0, 0, 1, 1])

# Trazar un gráfico de líneas utilizando la columna 'date' como eje X y 'temperatura' como eje Y, con un color verde
eje.plot(df['date'], df['temperatura'], color='g')

# Establecer los límites del eje X para mostrar datos en el intervalo deseado
eje.set_xlim(datetime.datetime(2013, 2, 1), datetime.datetime(2015, 1, 1))

# Establecer un título para el gráfico con un tamaño de fuente personalizado
eje.set_title('Temperatura en el tiempo', fontsize=25)

# Establecer etiquetas para los ejes Y y X con tamaños de fuente personalizados
eje.set_ylabel('Temperatura', fontsize=20)  # Etiqueta del eje Y
eje.set_xlabel('Años', fontsize=20)        # Etiqueta del eje X

# Agregar una leyenda para identificar la serie de datos
eje.legend(['temperatura'], loc='lower right', fontsize=15);
```

 * La importación de la librería datetime es importante en este caso porque necesitas trabajar con fechas y horas en Python, especialmente al configurar límites en los ejes de tiempo en tu gráfico. La librería datetime proporciona clases y funciones para trabajar con fechas, horas y operaciones relacionadas.
 * **eje.set_xlim(datetime.datetime(2014, 1, 1), datetime.datetime(2015, 1, 1))** se utiliza para establecer los límites del eje X (horizontal) del gráfico.
```python
# Crear una figura con un tamaño específico
fig = plt.figure(figsize=(15, 8))

# Agregar un conjunto de ejes que cubre toda la figura
eje = fig.add_axes([0, 0, 1, 1])

# Habilitar la cuadrícula en el gráfico
eje.grid(True)

# Traza una serie de datos de temperatura en función de la fecha en color verde ('g')
eje.plot(df['date'], df['temperatura'], color='g')

# Configuración del título del gráfico con tamaño de fuente y espaciado adicional
eje.set_title('Temperatura en el Tiempo', fontsize=25, pad=20)

# Agregar una leyenda al gráfico con la etiqueta 'temperatura' en la esquina inferior derecha
eje.legend(['temperatura'], loc='lower right', fontsize=15)

# Etiqueta del eje Y que indica "Temperatura" con un tamaño de fuente
eje.set_ylabel('Temperatura', fontsize=20)

# Etiqueta del eje X que indica "Años" con un tamaño de fuente
eje.set_xlabel('Años', fontsize=20)

# Identificar las coordenadas del punto máximo de temperatura y agregar la anotación "Maximo"
x1_max = df['date'][df['temperatura'].idxmax()]
y1_max = max(df['temperatura'])
x2_max = df['date'][df['temperatura'].idxmax() - 7000]
y2_max = max(df['temperatura']) - 5
eje.annotate('Máximo', xy=(x1_max, y1_max), fontsize=20, xytext=(x2_max, y2_max), arrowprops=dict(facecolor='k'))

# Identificar las coordenadas del punto mínimo de temperatura y agregar la anotación "Minimo"
x1_min = df['date'][df['temperatura'].idxmin()]
y1_min = min(df['temperatura'])
x2_min = df['date'][df['temperatura'].idxmin() - 7000]
y2_min = min(df['temperatura']) + 5
eje.annotate('Mínimo', xy=(x1_min, y1_min), fontsize=20, xytext=(x2_min, y2_min), arrowprops=dict(facecolor='k'))

# Agregar líneas de referencia horizontal en la temperatura máxima y mínima
eje.axhline(max(df['temperatura']), color='k', linestyle='--')
eje.axhline(min(df['temperatura']), color='k', linestyle='--')

# Mostrar el gráfico
plt.show()
```

En este caso, el parámetro **autopct** se establece en **'%.1f%%'**, lo que significa que se mostrará el porcentaje con un decimal en cada segmento del gráfico de pastel. Esto proporciona información adicional sobre la distribución de la temperatura por día de la semana.

La lista **explodir** contiene valores que indican cuánto separar cada segmento del centro del gráfico. Esto puede ser útil cuando se desea enfocar la atención en ciertos elementos del gráfico de pastel.

```python
# Crear una figura y un eje
fig = plt.figure(figsize=(5, 4))
eje = fig.add_axes([0, 0, 1, 1])

# Definir una lista de valores para la explosión de los segmentos
explodir = [0.1, 0, 0.1, 0, 0, 0.1, 0]

# Crear el gráfico de pastel con los datos de temperatura_por_dia_de_semana
# El parámetro 'explode' especifica cuánto separar cada segmento del centro
# El parámetro 'shadow=True' agrega sombreado al gráfico
eje.pie(temperatura_por_dia_de_semana, labels=temperatura_por_dia_de_semana.index, autopct='%.1f%%', explode=explodir, shadow=True)

# Configurar el título del gráfico
eje.set_title('Temperatura por día de la semana', fontsize=15, pad=10)

# Mostrar el gráfico
plt.show()
```

El sombreado (**shadow=True**) agrega una apariencia tridimensional al gráfico de pastel, lo que puede hacer que los segmentos resalten visualmente y proporcionar una sensación de profundidad. Puedes ajustar el valor **0.1** en la lista **explodir** para controlar cuánto separar ciertos segmentos del centro del gráfico y resaltarlos aún más.

###  <span style="color:lightgreen"> **Diagramas de caja** </span>

Un boxplot (también conocido como diagrama de caja o gráfico de caja y bigotes) es un tipo de gráfico estadístico utilizado para visualizar la distribución de un conjunto de datos. El boxplot proporciona información sobre la ubicación, dispersión, simetría y presencia de valores atípicos en los datos de una manera eficiente y fácil de entender. Aquí se describen las partes clave de un boxplot y su utilidad:

 <span style="color:orange"> **Las partes de un boxplot:** </span>

1. **Caja (box):** Representa el rango intercuartil (IQR), que es la distancia entre el primer cuartil (Q1) y el tercer cuartil (Q3) de los datos. La caja generalmente se dibuja desde Q1 hasta Q3 y contiene el 50% central de los datos. La altura de la caja muestra la dispersión intercuartil.

2. **Línea (whisker) superior e inferior (bigotes):** Los bigotes se extienden desde los bordes de la caja hacia arriba y hacia abajo. Los valores ubicados a una cierta distancia (a menudo 1.5 veces el IQR) desde Q1 o Q3 se consideran valores atípicos y se muestran como puntos separados.

3. **Línea central (mediana):** En el interior de la caja, se encuentra una línea que representa la mediana de los datos, que es el valor que divide el conjunto de datos en dos partes iguales.

4. **Puntos atípicos (outliers):** Los valores atípicos se muestran como puntos individuales fuera de los bigotes. Estos son valores que se consideran inusuales o extremos en comparación con el resto de los datos.

 <span style="color:orange"> **La utilidad de un boxplot:** </span>

* Proporciona una representación visual de la distribución de los datos, lo que facilita la identificación de tendencias y características importantes en los datos.

* Permite comparar fácilmente múltiples grupos o categorías de datos.

* Ayuda a detectar valores atípicos, que son puntos que se desvían significativamente de la mayoría de los datos.

* Muestra la mediana, cuartiles y la dispersión intercuartil, lo que proporciona información sobre la ubicación y la dispersión de los datos.

* Es una herramienta útil para resumir y presentar datos en informes y presentaciones.

###  <span style="color:lightgreen"> **Gráficos de histograma** </span>


Un histograma es un tipo de gráfico que se utiliza para representar la distribución de frecuencias de un conjunto de datos. En un histograma, los datos se dividen en "bins" o intervalos de valores, y se cuenta cuántas veces caen los datos en cada intervalo. Luego, se representa gráficamente el recuento de frecuencias en forma de barras, donde la altura de cada barra corresponde a la cantidad de datos en ese intervalo.

Los histogramas son útiles para visualizar la forma y la dispersión de una distribución de datos. Algunos de los usos más comunes de los histogramas incluyen:

1. **Identificar patrones de distribución:** Los histogramas pueden ayudar a identificar si los datos siguen una distribución normal (en forma de campana) o si presentan sesgos, asimetrías o múltiples modas (valores más frecuentes).

2. **Evaluar la dispersión:** Los histogramas pueden indicar cuán dispersos o concentrados están los datos en torno a ciertos valores.

3. **Detectar valores atípicos:** Los valores que caen muy lejos de la mayoría de los datos pueden ser identificados en un histograma como "colas" inusuales.

4. **Comparar distribuciones:** Los histogramas permiten comparar visualmente varias distribuciones de datos, lo que es útil en análisis comparativos.

5. **Resumir datos:** Los histogramas resumen la información de manera eficiente y proporcionan una visión rápida de la estructura de los datos.

 <span style="color:orange"> **Agragando anotaciones y lineas horizontales** </span>

```python
# Crear una figura y un eje en blanco
fig = plt.figure(figsize=(5,4))
eje = fig.add_axes([0,0,1,1])

# Calcular la media (mu) y la desviación estándar (sigma) de los datos de 'PetalLengthCm'
mu, sigma = df['PetalLengthCm'].mean(), df['PetalLengthCm'].std()

# Crear un histograma de los datos de 'PetalLengthCm' con 20 contenedores
eje.hist(df['PetalLengthCm'], bins=20)

# Establecer el título del gráfico
eje.set_title('Histograma', fontsize=15, pad=10)

# Etiquetar el eje x
eje.set_xlabel('PetalLengthCm', fontsize=15)

# Activar la cuadrícula en el gráfico
eje.grid(True)

# Agregar una anotación con los valores de la media (mu) y la desviación estándar (sigma)
eje.annotate('$mu = {0:.2f}$\n$sigma = {1:.2f}$'.format(mu, sigma), xy=(4, 10), fontsize=20);
```

La anotación en el gráfico muestra los valores de la media y la desviación estándar con dos decimales de precisión en las coordenadas (4.5, 20) en el gráfico. Esto proporciona información estadística relevante sobre los datos representados en el histograma.
